# 🏨 Hotel Booking Cancellation Prediction

## 📌 Problem Statement

Hotel booking cancellations have become a significant challenge for hotels around the world. Every canceled reservation results in financial losses, inefficient room allocation, and difficulties in resource planning. Therefore, predicting whether a booking will be canceled before the customer's arrival can help hotels optimize their operations and improve revenue management.

---

## 🎯 Project Objective

The main objective of this project is to build a Machine Learning Classification model capable of predicting whether a hotel reservation will be canceled based on customer information, booking details, and reservation history.

---

## 📊 Machine Learning Task

**Classification**

The target variable is:

**is_canceled**

- **0** → Booking Not Cancelled
- **1** → Booking Cancelled

# **Cancellation Rate by Month**

```python
# ============================================================
# Cancellation Rate by Month
# ============================================================

monthly_result = (
    df.groupby("arrival_date_month")
      .agg(
          total_bookings=("is_canceled","count"),
          canceled_bookings=("is_canceled","sum")
      )
      .reset_index()
)

monthly_result["cancel_rate_pct"] = round(
    monthly_result["canceled_bookings"] * 100
    / monthly_result["total_bookings"],
    2
)

month_order = [

    "January","February","March","April",
    "May","June","July","August",
    "September","October","November","December"

]

monthly_result["arrival_date_month"] = pd.Categorical(

    monthly_result["arrival_date_month"],

    categories=month_order,

    ordered=True

)

monthly_result = monthly_result.sort_values("arrival_date_month")

overall_cancel_rate = round(df["is_canceled"].mean()*100,2)

plt.figure(figsize=(12,6))

sns.barplot(

    data=monthly_result,

    x="arrival_date_month",

    y="cancel_rate_pct",

    palette="Reds_r"

)

plt.axhline(

    y=overall_cancel_rate,

    color="black",

    linestyle="--",

    label=f"Average = {overall_cancel_rate}%"

)

plt.title(

    f"Cancellation Rate by Month (Overall = {overall_cancel_rate}%)",

    fontsize=16

)

plt.xlabel("Arrival Month")

plt.ylabel("Cancellation Rate (%)")

plt.xticks(rotation=45)

plt.legend()

plt.tight_layout()

plt.show()
```

# **Cancellation Rate by Lead Time**

```python
# ============================================================
# Cancellation Rate by Lead Time
# ============================================================

bins = [0, 7, 30, 90, 180, 1000]

labels = [

    "0-7 Days",
    "8-30 Days",
    "31-90 Days",
    "91-180 Days",
    "180+ Days"

]

df["lead_time_bucket"] = pd.cut(

    df["lead_time"],

    bins=bins,

    labels=labels

)

leadtime_result = (

    df.groupby(

        "lead_time_bucket",

        observed=True

    )

    .agg(

        total_bookings=("is_canceled", "count"),

        canceled_bookings=("is_canceled", "sum")

    )

    .reset_index()

)

leadtime_result["cancel_rate_pct"] = round(

    leadtime_result["canceled_bookings"] * 100

    / leadtime_result["total_bookings"],

    2

)

display(leadtime_result)
```

```python
# ============================================================
# Cancellation Rate by Lead Time
# ============================================================

overall_cancel_rate = round(

    df["is_canceled"].mean() * 100,

    2

)

plt.figure(figsize=(10,6))

sns.barplot(

    data=leadtime_result,

    x="lead_time_bucket",

    y="cancel_rate_pct",

    hue="lead_time_bucket",

    palette="Blues_r",

    legend=False

)

plt.axhline(

    y=overall_cancel_rate,

    color="red",

    linestyle="--",

    linewidth=2,

    label=f"Overall Average = {overall_cancel_rate}%"

)

plt.title(

    f"Cancellation Rate by Lead Time (Overall = {overall_cancel_rate}%)",

    fontsize=16

)

plt.xlabel("Days Between Booking and Arrival")

plt.ylabel("Cancellation Rate (%)")

plt.xticks(rotation=15)

plt.legend()

plt.tight_layout()

plt.show()
```

# **Observation:**

 • The cancellation rate increases as the lead time becomes longer.

 • Reservations made more than 180 days before arrival have the highest cancellation rate.

 • Last-minute bookings (0–7 days) have the lowest cancellation rate.

 • This indicates that customers who book far in advance are more likely to change or cancel their plans.

# **Recommendation**

 The hotel should consider stricter cancellation policies or partial deposits for bookings made more than 30 days before arrival, as these bookings have significantly higher cancellation rates.

### ⚖️ Handling Class Imbalance

The target `is_canceled` is imbalanced (~63% Not Canceled vs ~37% Canceled after cleaning).
Left untreated, this biases models toward the majority class and caps **recall** on the
cancellation class — i.e. real cancellations get missed.

Instead of resampling the data (SMOTE/undersampling), which can distort the joint
distribution of a 74k-row real-world booking dataset, class imbalance is handled at the
**model level** using each algorithm's native class-weighting mechanism:

- `class_weight="balanced"` for Logistic Regression, Decision Tree, Random Forest, Extra Trees
- `scale_pos_weight` (negative/positive ratio) for XGBoost
- `class_weight="balanced"` for LightGBM
- `auto_class_weights="Balanced"` for CatBoost

This keeps every training example, avoids synthetic data, and directly tells each model to
penalize mistakes on the minority (canceled) class more heavily — which should raise recall
on `is_canceled == 1` compared to the unweighted baseline.


```python
# ============================================================
# Class Imbalance — compute weighting ratio for models that need it
# (e.g. XGBoost's scale_pos_weight)
# ============================================================

class_counts = y_train.value_counts()

scale_pos_weight = class_counts[0] / class_counts[1]

print("Class distribution in y_train:")
print(class_counts)
print()
print(f"scale_pos_weight (neg/pos) = {scale_pos_weight:.4f}")
```

# **Part 11.4 — Handle Missing Values**

```python
# ============================================================
# Part 11.4 : Missing Values
# ============================================================

from sklearn.impute import SimpleImputer

numeric_imputer = SimpleImputer(

    strategy="median"

)

categorical_imputer = SimpleImputer(

    strategy="most_frequent"

)

X_train[numeric_cols] = numeric_imputer.fit_transform(

    X_train[numeric_cols]

)

X_test[numeric_cols] = numeric_imputer.transform(

    X_test[numeric_cols]

)

X_train[categorical_cols] = categorical_imputer.fit_transform(

    X_train[categorical_cols]

)

X_test[categorical_cols] = categorical_imputer.transform(

    X_test[categorical_cols]

)

print("Missing Values in Train")

print(X_train.isnull().sum().sum())

print()

print("Missing Values in Test")

print(X_test.isnull().sum().sum())
```

# **Part 11.5 — Smart Encoding**

```python
"""
✅ hotel → LabelEncoder
✅ arrival_date_month → Ordinal Encoding
✅ meal → LabelEncoder
✅ country → Frequency Encoding
✅ market_segment → LabelEncoder
✅ distribution_channel → LabelEncoder
✅ reserved_room_type → LabelEncoder
✅ assigned_room_type → LabelEncoder
✅ deposit_type → LabelEncoder
✅ customer_type → LabelEncoder
"""
# ============================================================
# Part 11.5 : Smart Encoding
# ============================================================

from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OrdinalEncoder

label_encoders = {}

# ------------------------------------------------
# 1. Hotel
# ------------------------------------------------

hotel_encoder = LabelEncoder()

X_train["hotel"] = hotel_encoder.fit_transform(X_train["hotel"])

X_test["hotel"] = hotel_encoder.transform(X_test["hotel"])

label_encoders["hotel"] = hotel_encoder

# ------------------------------------------------
# 2. Arrival Month (Ordinal Encoding)
# ------------------------------------------------

months = [[

    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"

]]

month_encoder = OrdinalEncoder(categories=months)

X_train["arrival_date_month"] = month_encoder.fit_transform(
    X_train[["arrival_date_month"]]
)

X_test["arrival_date_month"] = month_encoder.transform(
    X_test[["arrival_date_month"]]
)

# ------------------------------------------------
# 3. Country (Frequency Encoding)
# ------------------------------------------------

country_frequency = X_train["country"].value_counts(normalize=True)

X_train["country"] = X_train["country"].map(country_frequency)

X_test["country"] = X_test["country"].map(country_frequency)

X_test["country"] = X_test["country"].fillna(0)

# ------------------------------------------------
# 4. Label Encoding
# ------------------------------------------------

label_columns = [

    "meal",

    "market_segment",

    "distribution_channel",

    "reserved_room_type",

    "assigned_room_type",

    "deposit_type",

    "customer_type"

]

for col in label_columns:

    encoder = LabelEncoder()

    # نجمع Train و Test قبل الـ fit
    all_values = pd.concat([
        X_train[col],
        X_test[col]
    ]).astype(str)

    encoder.fit(all_values)

    X_train[col] = encoder.transform(
        X_train[col].astype(str)
    )

    X_test[col] = encoder.transform(
        X_test[col].astype(str)
    )

    label_encoders[col] = encoder

print("="*60)
print("Encoding Finished Successfully")
print("="*60)
```

# **Part 12.3 — Evaluation Function**

```python
# ============================================================
# Part 12.3 : Evaluate Model
# ============================================================

results = []

def evaluate_model(model, name):

    start = time.time()

    # Train Model
    model.fit(X_train, y_train)

    # Predictions
    train_pred = model.predict(X_train)

    test_pred = model.predict(X_test)

    # Probabilities
    if hasattr(model, "predict_proba"):

        test_prob = model.predict_proba(X_test)[:, 1]

    else:

        test_prob = model.decision_function(X_test)

    end = time.time()

    # Metrics
    train_acc = accuracy_score(y_train, train_pred)

    test_acc = accuracy_score(y_test, test_pred)

    precision = precision_score(y_test, test_pred)

    recall = recall_score(y_test, test_pred)

    f1 = f1_score(y_test, test_pred)

    roc = roc_auc_score(y_test, test_prob)

    cv = cross_val_score(

        model,

        X_train,

        y_train,

        cv=5,

        scoring="accuracy",

        n_jobs=-1

    ).mean()

    results.append({

        "Model": name,

        "Train Accuracy": round(train_acc,4),

        "Test Accuracy": round(test_acc,4),

        "CV Score": round(cv,4),

        "Precision": round(precision,4),

        "Recall": round(recall,4),

        "F1 Score": round(f1,4),

        "ROC AUC": round(roc,4),

        "Training Time": round(end-start,2)

    })

    return model
```

# **Part 12.4 — Train All Models**

```python
from tqdm.auto import tqdm

results = []

trained_models = {}

pbar = tqdm(models.items(), total=len(models))

for name, model in pbar:

    pbar.set_description(f"Training {name}")

    trained_model = evaluate_model(model, name)

    trained_models[name] = trained_model

print("\nTraining Finished Successfully.")
```

instead of raw accuracy keeps that trade-off in view when picking the "best" model, instead
of silently favoring whichever model happens to guess the majority class most often.
